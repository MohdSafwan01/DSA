```python
marks = [('Safwan',200),('Anas',258),('Imaad',400)]
```

```python
def marks_check(marks):

  current_max=0
  top_scorrer= ''

  for scorrer,score in marks:
    if score > current_max:
      current_max = score
      top_scorrer = scorrer
    else:
      pass

  return (top_scorrer,current_max)

```

```python
result = marks_check(marks)
```

LESSER OF TWO EVENS: Write a function that returns the lesser of two given numbers if both numbers are even, but returns the greater if one or both numbers are odd

```python
def lesser_of_two_evens(a,b):
  if a%2==0 and b%2==0:
    return min(a,b)
  else:
    return max(a,b)
  pass
```

```python
lesser_of_two_evens(5,7)
```


ANIMAL CRACKERS: Write a function takes a two-word string and returns True if both words begin with same letter

```python
def animal_cracker(text):
  word = text.split()
  if word[0][0]==word[1][0]:
    return True
  else:
    return False

```

```python
animal_cracker("Levelheaded Llamma")
```

```python
animal_cracker("crazy kangaroo")
```


MAKES TWENTY: Given two integers, return True if the sum of the integers is 20 or if one of the integers is 20. If not, return False

```python
def makes_twenty(n1,n2):
  if n1 == 20 or n2 == 20:
    return True
  elif n1+n2==20:
    return True
  else:
    return False
```

```python
makes_twenty(20,3)
```

MASTER YODA: Given a sentence, return a sentence with the words reversed
master_yoda('I am home') --> 'home am I'
master_yoda('We are ready') --> 'ready are We'

```python
def master_yoda(text):
  new_text = text.split()
  final_text=new_text[::-1]
  Answer_text=" ".join(final_text)
  return Answer_text
```

```python
master_yoda('I am Home')
```

```python
master_yoda('my name is Safwan')
```

ALMOST THERE: Given an integer n, return True if n is within 10 of either 100 or 200

NOTE: abs(num) returns the absolute value of a number
